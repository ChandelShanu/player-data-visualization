```python
def mostsus(df,sports):
    
    temp=df.dropna(subset=["Medal"])
    if sports!= 'Overall':
       temp=temp[temp["Sport"]==sports]
    return temp['Name'].value_counts().reset_index().head(14).merge(df,on='Name',how='left')[["Name",'count','Sport','region']].drop_duplicates("Name")
```
